```python
# PROGRAM PENYELESAIAN SPL 3 VARIABEL
# MENGGUNAKAN METODE ELIMINASI GAUSS

import numpy as np

print("PROGRAM SPL 3 VARIABEL")
print("METODE ELIMINASI GAUSS")
```

```python
# Memasukkan koefisien persamaan ke dalam matriks

A = np.array([
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
], dtype=float)

B = np.array([8, -11, -3], dtype=float)

print("Matriks koefisien:")
print(A)

print("Hasil persamaan:")
print(B)
```

```python
# PROSES ELIMINASI GAUSS

# Menggabungkan matriks koefisien dan hasil
M = np.column_stack((A, B))

# Menentukan jumlah variabel
n = len(B)

# Melakukan eliminasi
for i in range(n):

    # Mencari posisi pivot terbesar
    pivot = i + np.argmax(abs(M[i:, i]))

    # Menukar posisi baris
    M[[i, pivot]] = M[[pivot, i]]

    # Menghilangkan angka di bawah pivot
    for j in range(i + 1, n):
        faktor = M[j, i] / M[i, i]
        M[j] = M[j] - faktor * M[i]

print("Matriks setelah eliminasi:")
print(M)
```

```python
# PROSES SUBSTITUSI MUNDUR

# Menyiapkan tempat untuk menyimpan hasil
x = np.zeros(n)

# Menghitung nilai variabel dari bawah ke atas
for i in range(n - 1, -1, -1):

    x[i] = (
        M[i, -1] -
        np.dot(M[i, i+1:n], x[i+1:n])
    ) / M[i, i]

# Menampilkan hasil perhitungan
print("HASIL PENYELESAIAN SPL 3 VARIABEL")
print("--------------------------------")
print("Nilai x =", round(x[0], 2))
print("Nilai y =", round(x[1], 2))
print("Nilai z =", round(x[2], 2))
```

### CELL 1 – Persiapan Program

Cell ini merupakan tahap awal program penyelesaian Sistem Persamaan Linear (SPL) tiga variabel menggunakan metode Eliminasi Gauss. Program menggunakan library NumPy untuk membantu melakukan operasi matematika, seperti membuat dan mengolah matriks. Perintah print() digunakan untuk menampilkan judul program kepada pengguna.

### CELL 2 – Memasukkan Matriks Persamaan

Pada bagian ini, pengguna menentukan koefisien dari tiga persamaan linear yang memiliki variabel x, y, dan z. Matriks A menyimpan koefisien setiap variabel, sedangkan matriks B menyimpan hasil dari masing-masing persamaan. Tipe data float digunakan agar perhitungan dapat menghasilkan bilangan desimal.

### CELL 3 – Proses Eliminasi Gauss

Cell ini digunakan untuk mengubah matriks persamaan menjadi bentuk segitiga atas. Pertama, matriks A dan B digabungkan menjadi matriks augmented. Selanjutnya program mencari elemen pivot, menukar baris apabila diperlukan, dan melakukan eliminasi untuk mengubah elemen di bawah diagonal utama menjadi nol.

### CELL 4 – Proses Substitusi Mundur

Pada tahap ini dilakukan substitusi mundur untuk mendapatkan nilai setiap variabel. Perhitungan dimulai dari variabel terakhir, kemudian dilanjutkan ke variabel sebelumnya hingga seluruh nilai x, y, dan z diperoleh. Fungsi round() digunakan untuk membatasi angka desimal agar hasil lebih mudah dibaca.
